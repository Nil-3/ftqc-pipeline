```python
import numpy as np
from scipy.optimize import curve_fit
noise_scales = np.array([1, 3, 5])
exp_values = np.array([0.85, 0.60, 0.40])
def fit_func(x, a, b):
 return a * x + b
params, _ = curve_fit(fit_func, noise_scales, exp_values)
mitigated_value = fit_func(0, *params)
print(f'Mitigated Expectation Value: {mitigated_value:.3f}')
```
