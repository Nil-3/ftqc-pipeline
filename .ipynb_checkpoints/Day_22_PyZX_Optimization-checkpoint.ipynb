```python
import pyzx as zx
import os

project_dir = r"C:\Users\USER\FTQC_Project"
os.chdir(project_dir)

def optimize_circuit(molecule_name):
    print(f"Optimizing {molecule_name}...")
    circuit = zx.Circuit.load(f"{molecule_name}_active_space.qasm")
    g = circuit.to_graph()
    zx.full_reduce(g)
    optimized = zx.extract_circuit(g.copy())
    t_count = optimized.tcount()
    
    with open(f"{molecule_name}_optimized.qasm", "w") as f:
        f.write(optimized.to_qasm())
    return t_count

# Running for all molecules
for mol in ["Water", "Carbon Dioxide", "Ammonia", "Aspirin"]:
    t = optimize_circuit(mol)
    print(f"Success! {mol} | Final T-count: {t}\n")
```
