```python
from qiskit.transpiler import CouplingMap
from qiskit.transpiler.preset_passmanagers import generate_preset_pass_manager
hex_edges = [[0,1], [1,2], [1,3], [3,4], [4,5], [4,6]]
hex_map = CouplingMap(hex_edges)
pass_manager = generate_preset_pass_manager(optimization_level=3, 
coupling_map=hex_map)
# optimized_hex_circ = pass_manager.run(your_circuit)
print("Heavy hex coupling map successfully generated")
print("Number of physical qubits in map:", hex_map.size())
```
