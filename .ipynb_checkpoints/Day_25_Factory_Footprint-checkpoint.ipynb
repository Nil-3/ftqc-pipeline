```python
import matplotlib.pyplot as plt

# Data points from Day 24 scaling analysis (T-count: 16)
distances = [11, 15, 19, 23]
latencies = [4.40, 6.00, 7.60, 9.20]        # in milliseconds
physical_qubits = [3368, 6600, 7688, 9032]   # total physical qubits

fig, ax1 = plt.subplots(figsize=(8, 5))

# Plotting Latency on primary y-axis
color = 'tab:red'
ax1.set_xlabel('Surface Code Distance (d)', fontsize=12)
ax1.set_ylabel('Parallel Latency (ms)', color=color, fontsize=12)
ax1.plot(distances, latencies, color=color, marker='o', linewidth=2, label='Latency')
ax1.tick_params(axis='y', labelcolor=color)
ax1.grid(True, linestyle='--', alpha=0.5)

# Creating a secondary y-axis for Physical Qubits
ax2 = ax1.twinx()  
color = 'tab:blue'
ax2.set_ylabel('Physical Qubit Footprint', color=color, fontsize=12)
ax2.plot(distances, physical_qubits, color=color, marker='s', linewidth=2, linestyle='--', label='Physical Qubits')
ax2.tick_params(axis='y', labelcolor=color)

plt.title('Fault-Tolerant Resource Trade-Off Analysis (T-count: 16)', fontsize=13, pad=15)
fig.tight_layout()

# Save high-resolution figure for your thesis/paper
plt.savefig('ftqc_scaling_tradeoff.png', dpi=300)
plt.show()
```
