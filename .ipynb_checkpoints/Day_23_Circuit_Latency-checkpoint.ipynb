```python
import os
import pyzx as zx

project_dir = r"C:\Users\USER\FTQC_Project"
os.chdir(project_dir)

def calculate_latency(t_count, distance_d=15):
    # Assume 1 microsecond per surface code cycle
    latency_ms = (t_count * distance_d) / 1000
    print(f"Calculated Latency: {latency_ms} ms")
    return latency_ms

# Running the latency analysis for all  Day 22 optimized files
target_molecules = ["Water", "Carbon Dioxide", "Ammonia", "Aspirin"]

for mol in target_molecules:
    optimized_file = f"{mol}_optimized.qasm"
    if os.path.exists(optimized_file):
        print(f"Analyzing {mol}...")
        circuit = zx.Circuit.load(optimized_file)
        t_count = circuit.tcount()
        print(f"  -> T-count: {t_count}")
        calculate_latency(t_count, distance_d=15)
        print()
    else:
        print(f"Could not find {optimized_file}\n")
```

```python
import os
import pyzx as zx

project_dir = r"C:\Users\USER\FTQC_Project"
os.chdir(project_dir)

def estimate_fault_tolerant_resources(t_count, distance_d=15, factory_cycle_time=50.0, num_factories=2):
    # 1. Parallel Factory Latency Calculation
    effective_cycles = t_count / num_factories
    latency_microseconds = effective_cycles * factory_cycle_time
    latency_milliseconds = latency_microseconds / 1000
    
    # 2. Surface Code Physical Qubit Estimation
    logical_qubits = 4  
    data_qubits_physical = logical_qubits * (2 * (distance_d ** 2))
    factory_qubits_physical = num_factories * 1200  # Overhead for distillation factories
    total_physical_qubits = data_qubits_physical + factory_qubits_physical
    
    return latency_milliseconds, total_physical_qubits

target_molecules = ["Water", "Carbon Dioxide", "Ammonia", "Aspirin"]

print("--- Fault-Tolerant Parallel Factory & Footprint Analysis ---\n")

for mol in target_molecules:
    optimized_file = f"{mol}_optimized.qasm"
    if os.path.exists(optimized_file):
        circuit = zx.Circuit.load(optimized_file)
        t_count = circuit.tcount()
        
        latency, physical_qubits = estimate_fault_tolerant_resources(t_count)
        
        print(f"Molecule: {mol}")
        print(f"  -> T-count: {t_count}")
        print(f"  -> Parallel Latency (2 factories): {latency:.4f} ms")
        print(f"  -> Est. Physical Qubits (d=15): {physical_qubits:,} qubits\n")
    else:
        print(f"Could not find {optimized_file}\n")
```
