```python
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister
# 9 data qubits, 8 measure qubits for d=3 surface code
data_q = QuantumRegister(9, 'data')
synd_q = QuantumRegister(8, 'syndrome')
synd_c = ClassicalRegister(8, 'synd_c')
qc_surface = QuantumCircuit(data_q, synd_q, synd_c)
# Example Z-stabilizer on qubits 0, 1, 3, 4
qc_surface.h(synd_q[0])
qc_surface.cx(data_q[0], synd_q[0])
qc_surface.cx(data_q[1], synd_q[0])
qc_surface.cx(data_q[3], synd_q[0])
qc_surface.cx(data_q[4], synd_q[0])
qc_surface.h(synd_q[0])
qc_surface.measure(synd_q[0], synd_c[0])
print('Surface Code Encoding Circuit Depth:', qc_surface.depth())
```
