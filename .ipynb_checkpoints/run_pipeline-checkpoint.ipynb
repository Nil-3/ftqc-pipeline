```python
import os
import pyzx as zx

project_dir = r"C:\Users\USER\FTQC_Project"
os.chdir(project_dir)

def calculate_total_overhead(data_qubits, factory_qubits):
    return data_qubits + factory_qubits

target_molecules = ["Water", "Carbon Dioxide", "Ammonia", "Aspirin"]
distances = [11, 15, 19, 23]
num_factories = 2

print("==================================================")
print("  FTQC MASTER RESOURCE ESTIMATION PIPELINE (DAY 27)")
print("==================================================\n")

for mol in target_molecules:
    optimized_file = f"{mol}_optimized.qasm"
    if os.path.exists(optimized_file):
        circuit = zx.Circuit.load(optimized_file)
        t_count = circuit.tcount()
        
        print(f"[*] Target Molecule: {mol}")
        print(f"    Optimized T-count: {t_count}")
        print(f"    {'Distance (d)':<12} | {'Data Qubits':<15} | {'Factory Qubits':<15} | {'Total Physical':<15}")
        print("    " + "-" * 61)
        
        for d in distances:
            logical_qubits = 4
            data_qubits = logical_qubits * (2 * (d ** 2))
            factory_qubits = num_factories * (1200 * (d // 15 + 1))
            total_physical = calculate_total_overhead(data_qubits, factory_qubits)
            
            print(f"    {d:<12} | {data_qubits:<15,} | {factory_qubits:<15,} | {total_physical:<15,}")
        print("\n" + "=" * 50 + "\n")
```
