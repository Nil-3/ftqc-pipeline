```python
# System throughput calculation
instructions_per_sec = 1e4
quantum_overhead = 1.15 # 15% routing overhead
effective_throughput = instructions_per_sec / quantum_overhead
print(f'Effective Coprocessor Throughput: {effective_throughput:.1f} ops/sec')
```
