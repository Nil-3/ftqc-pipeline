```python
import pyzx as zx

circ = zx.Circuit.from_qasm_file("trotter_benchmark_ftqc.qasm")
print(f"Initial Circuit Gates: {len(circ.gates)}")
print(f"Initial T-count: {circ.tcount()}")
```

```python
# 2. Converting the quantum circuit into a ZX-calculus graph
g = circ.to_graph()
# 3. Apply the full reduction pipeline (Spider fusion, Phase gadget simplification)
# This algorithm eliminates redundant phases, massively reducing T-count
zx.full_reduce(g)
# Making the graph suitable for display and draw it
g.normalize()
zx.draw(g)
```

```python
optimized_circ = zx.extract_circuit(g.copy())

print(f"Optimized Circuit Gates: {len(optimized_circ.gates)}")
print(f"Optimized T-count: {optimized_circ.tcount()}")
```

```python
# 4. Extract the equivalent quantum circuit from the simplified ZX-diagram
circ_opt = zx.extract_circuit(g.copy())
# 5. Apply basic gate-level optimization to clean up the extracted circuit
circ_opt = zx.optimize.basic_optimization(circ_opt.to_basic_gates())
# 6. Compare the results
t_reduction = 100 * (circ.tcount() - circ_opt.tcount()) / max(circ.tcount(), 1)
print("="*45)
print("ZX-CALCULUS OPTIMIZATION REPORT")
print("="*45)
print(f"Optimized Circuit Gates: {len(circ_opt.gates)}")
print(f"Optimized T-count : {circ_opt.tcount()}")
print(f"T-count Reduction : {t_reduction:.2f}%")
print("="*45)
```
