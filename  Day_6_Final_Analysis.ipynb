```python
import matplotlib.pyplot as plt

metrics = {
 'baseline_t_count': 150, 
 'optimized_t_count': 95,
 'logical_depth': 40, 
 'physical_depth': 75 
}
print("Metrics loaded successfully for visualization.")
```

```python
labels = ['Baseline (Day 3)', 'ZX Optimized (Day 4)']
t_counts = [metrics['baseline_t_count'], metrics['optimized_t_count']]
plt.figure(figsize=(6, 4))
bars = plt.bar(labels, t_counts, color=['#EF4444', '#22C55E'])
plt.title('Logical T-Count Reduction via ZX-Calculus', fontweight='bold')
plt.ylabel('Number of T-Gates')
plt.grid(axis='y', linestyle='--', alpha=0.7)
# Add value labels on top of the bars
for bar in bars:
 yval = bar.get_height()
 plt.text(bar.get_x() + bar.get_width()/2, yval + 2, str(yval), ha='center', fontweight='bold')
plt.show()
```

```python
labels_depth = ['Logical Layout', 'Physical Layout (Day 5)']
depths = [metrics['logical_depth'], metrics['physical_depth']]
plt.figure(figsize=(6, 4))
bars = plt.bar(labels_depth, depths, color=['#3B82F6', '#F59E0B'])
plt.title('Circuit Depth: Logical vs Physical Hardware Mapping', fontweight='bold')
plt.ylabel('Circuit Depth (Time Steps)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
for bar in bars:
 yval = bar.get_height()
 plt.text(bar.get_x() + bar.get_width()/2, yval + 2, str(yval), ha='center', fontweight='bold')
plt.show()
```
