```python
import matplotlib.pyplot as plt
logical_qubits = [10, 20, 50, 100]
physical_qubits = [q * 17 for q in logical_qubits] # roughly d=3 overhead
plt.plot(logical_qubits, physical_qubits, marker='o', color='purple')
plt.title('Logical vs Physical Qubit Scaling')
plt.xlabel('Logical Qubits')
plt.ylabel('Physical Qubits Required')
plt.grid(True)
plt.savefig('scalability_plot.png')
print('Saved scalability plot!')
```
