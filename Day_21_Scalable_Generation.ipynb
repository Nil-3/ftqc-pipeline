```python
import os
import pyzx as zx

project_dir = r"C:\Users\USER\FTQC_Project"
os.chdir(project_dir)

target_molecules = ["Water", "Carbon Dioxide", "Ammonia", "Aspirin"]

print("Starting local FTQC optimization pipeline...\n")

for mol in target_molecules:
    active_space_file = f"{mol}_active_space.qasm"
    
    if os.path.exists(active_space_file):
        print(f"Optimizing {mol}...")
        
        # Loading the circuit generated from Colab
        zx_c = zx.Circuit.load(active_space_file)
        g = zx_c.to_graph()
    
        # Optimize with PyZX
        zx.full_reduce(g)
        opt_c = zx.extract_circuit(g)
        
        # Saving the new optimized circuit locally
        optimized_file = f"{mol}_optimized.qasm"
        with open(optimized_file, "w") as f:
            f.write(opt_c.to_qasm())
            
        t_count = opt_c.tcount()
        print(f"Success! {mol} optimized | Final T-count: {t_count}\n")
    else:
        print(f"Error: Could not find '{active_space_file}'. Double-check that it is in the FTQC_Project folder.\n")
```
