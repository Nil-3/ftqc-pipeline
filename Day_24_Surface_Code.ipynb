```python
import os
import pyzx as zx

project_dir = r"C:\Users\USER\FTQC_Project"
os.chdir(project_dir)

def analyze_distance_scaling(t_count, distances=[11, 15, 19, 23], factory_cycle_time=50.0, num_factories=2):
    results = []
    for d in distances:
        # Latency scales with distance because surface code cycle time depends on d
        latency_ms = ((t_count / num_factories) * factory_cycle_time * d) / 1000
        
        # Physical qubits scale quadratically with distance (d^2)
        logical_qubits = 4
        data_qubits = logical_qubits * (2 * (d ** 2))
        factory_qubits = num_factories * (1200 * (d // 15 + 1)) # scaling factor approximation
        total_qubits = data_qubits + factory_qubits
        
        results.append((d, latency_ms, total_qubits))
    return results

target_molecules = ["Water", "Carbon Dioxide", "Ammonia", "Aspirin"]

print("--- Day 24: Fault-Tolerant Code Distance Scaling Analysis ---\n")

for mol in target_molecules:
    optimized_file = f"{mol}_optimized.qasm"
    if os.path.exists(optimized_file):
        circuit = zx.Circuit.load(optimized_file)
        t_count = circuit.tcount()
        
        print(f"Molecule: {mol} (T-count: {t_count})")
        scaling_results = analyze_distance_scaling(t_count)
        
        print(f"  {'Distance (d)':<12} | {'Latency (ms)':<15} | {'Physical Qubits':<15}")
        print("-" * 48)
        for d, lat, q_count in scaling_results:
            print(f"  {d:<12} | {lat:<15.2f} | {q_count:<15,}")
        print()
```
