```python
import os
import zipfile
import glob

# Day 30: Clean Bundle Archive Script (Excluding Legacy LiH Files)
zip_filename = "ftqc_thesis_project_bundle.zip"

# Explicitly defined core deliverables
files_to_bundle = [
    "run_pipeline.ipynb",
    "true_hardware_scaling.png",
    "thesis_final_metrics.csv",
    "presentation_manifest.json"
]

# Dynamically gather QASM files, filtering out any legacy/unwanted files (like 'lih')
qasm_files = glob.glob("*.qasm")
filtered_qasm = [f for f in qasm_files if not f.startswith("lih_")]
files_to_bundle.extend(filtered_qasm)

with zipfile.ZipFile(zip_filename, 'w', zipfile.ZIP_DEFLATED) as zipf:
    for file in files_to_bundle:
        if os.path.exists(file):
            zipf.write(file)
            print(f"[+] Added to archive: {file}")
        else:
            print(f"[-] Skipped (not found): {file}")

print(f"\n[SUCCESS] Clean project bundle saved as '{zip_filename}' (Legacy LiH files excluded).")
```
